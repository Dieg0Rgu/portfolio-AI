```python
import os
from dotenv import load_dotenv
from google import genai as gemini

# 1. Cargamos las variables de entorno del archivo .env
load_dotenv()

# 2. Inicializamos el cliente (busca automáticamente la variable GEMINI_API_KEY)
# Asegúrate de que en tu archivo .env la línea se llame exactamente: GEMINI_API_KEY=tu_clave_aqui
cliente = gemini.Client() 

prompt = f"""
Haz un análisis del último año de las acciones de las empresas Apple, Microsoft y Nvidia considerando eventos y noticias que justifiquen las alzas y bajas relevantes de estas empresas.

{precios}
"""

# Usando el método recomendado para evitar advertencias de AFC
chat = cliente.chats.create(model="gemini-3.5-flash")
respuesta = chat.send_message(prompt)

# Para ver el resultado final en tu Jupyter Notebook:
print(respuesta.text)
```
