```python
import pandas as pd
import plotly.express as px

# Cargar los datos de ventas del archivo Excel.
ruta_archivo = "datos_ejemplo.xlsx"
datos = pd.read_excel(ruta_archivo)

# Calcular el total de ventas y el promedio de ventas diarias.
total_ventas = datos["Ventas"].sum()
promedio_ventas_diarias = datos["Ventas"].mean()

print(f"Total de ventas: {total_ventas:,.2f}")
print(f"Promedio de ventas diarias: {promedio_ventas_diarias:,.2f}")

# Agrupar las ventas por categoría y crear la gráfica de barras.
ventas_por_categoria = (
    datos.groupby("Categoria", as_index=False)["Ventas"]
    .sum()
    .sort_values("Ventas", ascending=False)
)

figura = px.bar(
    ventas_por_categoria,
    x="Categoria",
    y="Ventas",
    title="Total de ventas por categoría",
    labels={"Categoria": "Categoría", "Ventas": "Total de ventas"},
    text_auto=".2s",
)
figura.update_layout(xaxis_title="Categoría", yaxis_title="Ventas")
figura.show()
```
