# Contador de Palabras con Python

```python
from collections import Counter
import re

# Texto de ejemplo
texto = """
Python es un lenguaje de programación poderoso.
Python es fácil de aprender y muy versátil.
Muchos programadores usan Python para desarrollo web.
"""

# Limpiar el texto y convertir a minúsculas
texto_limpio = texto.lower()

# Extraer palabras usando expresiones regulares
palabras = re.findall(r'\b[a-záéíóúñ]+\b', texto_limpio)

# Contar las palabras
contador_palabras = Counter(palabras)

# Resultados
print(f"Total de palabras: {len(palabras)}")
print(f"Palabras únicas: {len(contador_palabras)}")
print(f"\nFrecuencia de palabras:")
for palabra, frecuencia in contador_palabras.most_common():
    print(f"  {palabra}: {frecuencia}")
```
