# Análisis de Venta

```python
# ==========================================
# Celda 1: Importación y Carga de Datos
# ==========================================
# Importar librerías necesarias
import pandas as pd
import plotly.express as px

# Cargar el archivo Excel
df = pd.read_excel('datos_ejemplo.xlsx')

# Mostrar las primeras filas para verificar la carga
print("Primeras filas del dataset:")
print(df.head())
print(f"\nDimensiones: {df.shape[0]} filas, {df.shape[1]} columnas")


# ==========================================
# Celda 2: Métricas Principales
# ==========================================
# Calcular métricas principales
total_ventas = df['Ventas'].sum()
promedio_ventas_diarias = df['Ventas'].mean()

print("=" * 50)
print("MÉTRICAS DE VENTAS 2024")
print("=" * 50)
print(f"Total de Ventas: ${total_ventas:,.2f}")
print(f"Promedio de Ventas Diarias: ${promedio_ventas_diarias:,.2f}")
print("=" * 50)


# ==========================================
# Celda 3: Gráfica de Barras con Plotly
# ==========================================
# Agrupar ventas por Categoría
ventas_por_categoria = df.groupby('Categoria')['Ventas'].sum().reset_index()
ventas_por_categoria = ventas_por_categoria.sort_values('Ventas', ascending=False)

# Crear gráfica de barras con Plotly
fig = px.bar(
    ventas_por_categoria,
    x='Categoria',
    y='Ventas',
    title='Total de Ventas por Categoría (2024)',
    labels={'Ventas': 'Total Ventas ($)', 'Categoria': 'Categoría'},
    text='Ventas',
    color='Categoria',  # (Nota: en la versión inicial estaba como color='Ventas' con color_continuous_scale='Viridis')
)

# Personalizar la gráfica
fig.update_traces(texttemplate='$%{text:,.0f}', textposition='outside')
fig.update_layout(
    xaxis_title='Categoría',
    yaxis_title='Total Ventas ($)',
    hovermode='x unified',
    height=500
)

fig.show()
```
