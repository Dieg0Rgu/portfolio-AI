```python
import yfinance as yf

tikers = ["AAPL", "NVDA", "MSFT"]


datos = yf.download(tikers,period = "1y")

precios = datos["Close"]
```

## Gráficos interactivos


```python
%pip install plotly
```

```python

import plotly.express as px
import plotly.io as pio

pio.renderers.default = "notebook_connected"

test= px.line(precios)
test.show()
```

```python
px.line(precios, 
    title="Cotización Histórica: Apple, Microsoft y Nvidia (últimos 12 meses)",
    labels={"value" : "Precio (USD)", "Date" : "Fechas"},
    template="plotly_dark")
    
```

```python
px.bar(precios.mean(),
       title = "Precio promedio de Apple, Nvidia y Microsoft (últimos 12 meses)",
       labels= {"value" : "Precio promedio (USD)", "Ticker" : "Empresa"},
       color=precios.mean().index,
       template="plotly_dark")
```
